# Dashboard Simples e Funcional

```python
import dash
from dash import dcc, html, Input, Output
import plotly.express as px
import pandas as pd
import numpy as np

print("✅ Imports OK")
```

```python
# Data
np.random.seed(42)
df = pd.DataFrame({
    'project': [f'Project {i}' for i in range(1, 16)],
    'status': np.random.choice(['Done', 'Progress', 'Hold'], 15),
    'completion': np.random.randint(30, 100, 15),
    'budget': np.random.randint(10000, 100000, 15)
})
print("✅ Data created")
df.head()
```

```python
# App
app = dash.Dash(__name__)

app.layout = html.Div([
    html.H1("Simple Dashboard", style={'textAlign': 'center'}),
    
    dcc.Graph(
        id='pie-chart',
        figure=px.pie(df['status'].value_counts(), 
                     values=df['status'].value_counts().values,
                     names=df['status'].value_counts().index,
                     title="Status Distribution")
    ),
    
    dcc.Graph(
        id='bar-chart', 
        figure=px.bar(df, x='project', y='completion', 
                     color='status', title="Project Progress")
    )
])

print("✅ App created")
```

```python
# Run
print("🚀 Starting on http://localhost:8053")
app.run_server(debug=False, port=8053, host='0.0.0.0')
```
